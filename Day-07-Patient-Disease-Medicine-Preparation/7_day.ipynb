```python
import pandas as pd
import numpy as np

from pathlib import Path
```

```python
input_file = Path("clinical_feature_engineered_data.xlsx")
output_file = Path("patient_disease_medicine_data.xlsx")

print("Input file:", input_file)
print("Output file:", output_file)
```

```python
if not input_file.exists():
    raise FileNotFoundError(
        f"Day 6 output file was not found: {input_file}"
    )

print("Day 6 file found successfully.")
```

```python
excel_file = pd.ExcelFile(input_file)

print("Available sheets:")
print(excel_file.sheet_names)
```

```python
df = pd.read_excel(
    input_file,
    sheet_name="Feature_Engineered_Data"
)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df.head()
```

```python
print("Dataset Information")
print("=" * 50)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nData Types:")
print(df.dtypes)
```

```python
patient_column = "patient_id"
disease_column = "condition"
medicine_column = "historical_medication_class"

required_columns = [
    patient_column,
    disease_column,
    medicine_column
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

print("Required columns verified successfully.")
print(required_columns)
```

```python
patient_disease_medicine = df[
    [
        patient_column,
        disease_column,
        medicine_column
    ]
].copy()

patient_disease_medicine.columns = [
    "patient_id",
    "disease",
    "medicine_class"
]

print("Patient–Disease–Medicine dataset shape:")
print(patient_disease_medicine.shape)
```

```python
patient_disease_medicine.head(10)
```

```python
missing_relationship_values = (
    patient_disease_medicine.isnull().sum()
)

print("Missing values:")
print(missing_relationship_values)
```

```python
blank_value_counts = (
    patient_disease_medicine
    .astype(str)
    .apply(lambda column: column.str.strip().eq("").sum())
)

print("Blank values:")
print(blank_value_counts)
```

```python
unique_patients = (
    patient_disease_medicine["patient_id"]
    .nunique()
)

print("Total records:", len(patient_disease_medicine))
print("Unique patients:", unique_patients)
```

```python
duplicate_relationships = (
    patient_disease_medicine
    .duplicated()
    .sum()
)

print("Duplicate Patient–Disease–Medicine relationships:")
print(duplicate_relationships)
```

```python
duplicate_rows = (
    patient_disease_medicine[
        patient_disease_medicine.duplicated(keep=False)
    ]
    .sort_values(
        ["patient_id", "disease", "medicine_class"]
    )
)

duplicate_rows.head(20)
```

```python
patient_disease_medicine_unique = (
    patient_disease_medicine
    .drop_duplicates()
    .reset_index(drop=True)
)

print(
    "Unique Patient–Disease–Medicine records:",
    len(patient_disease_medicine_unique)
)
```

```python
disease_medicine_map = (
    patient_disease_medicine_unique[
        ["disease", "medicine_class"]
    ]
    .drop_duplicates()
    .sort_values(
        ["disease", "medicine_class"]
    )
    .reset_index(drop=True)
)

print("Disease–Medicine mapping:")
print(disease_medicine_map)
```

```python
disease_medicine_counts = (
    patient_disease_medicine_unique
    .groupby(
        ["disease", "medicine_class"],
        as_index=False
    )
    .agg(
        patient_count=("patient_id", "nunique")
    )
    .sort_values(
        "patient_count",
        ascending=False
    )
    .reset_index(drop=True)
)

disease_medicine_counts.head(20)
```

```python
number_of_diseases = (
    patient_disease_medicine_unique["disease"]
    .nunique()
)

print("Number of unique diseases/conditions:")
print(number_of_diseases)
```

```python
number_of_medicine_classes = (
    patient_disease_medicine_unique["medicine_class"]
    .nunique()
)

print("Number of unique medication classes:")
print(number_of_medicine_classes)
```

```python
patient_disease = (
    patient_disease_medicine_unique[
        ["patient_id", "disease"]
    ]
    .drop_duplicates()
    .sort_values(
        ["patient_id", "disease"]
    )
    .reset_index(drop=True)
)

print("Patient–Disease table shape:")
print(patient_disease.shape)
```

```python
patient_count_by_disease = (
    patient_disease
    .groupby("disease", as_index=False)
    .agg(
        patient_count=("patient_id", "nunique")
    )
    .sort_values(
        "patient_count",
        ascending=False
    )
    .reset_index(drop=True)
)

patient_count_by_disease
```

```python
total_unique_patients = (
    patient_disease["patient_id"]
    .nunique()
)

patient_count_by_disease["patient_percentage"] = (
    patient_count_by_disease["patient_count"]
    / total_unique_patients
    * 100
)

patient_count_by_disease[
    "patient_percentage"
] = patient_count_by_disease[
    "patient_percentage"
].round(2)

patient_count_by_disease
```

```python
disease_summary = (
    patient_disease_medicine_unique
    .groupby("disease")
    .agg(
        patient_count=("patient_id", "nunique"),
        medicine_class_count=("medicine_class", "nunique")
    )
    .reset_index()
    .sort_values(
        "patient_count",
        ascending=False
    )
    .reset_index(drop=True)
)

disease_summary
```

```python
medicine_class_summary = (
    patient_disease_medicine_unique
    .groupby("medicine_class")
    .agg(
        patient_count=("patient_id", "nunique"),
        disease_count=("disease", "nunique")
    )
    .reset_index()
    .sort_values(
        "patient_count",
        ascending=False
    )
    .reset_index(drop=True)
)

medicine_class_summary
```

```python
medicine_distribution = (
    patient_disease_medicine_unique[
        "medicine_class"
    ]
    .value_counts()
    .rename_axis("medicine_class")
    .reset_index(name="patient_count")
)

medicine_distribution
```

```python
medicine_distribution["percentage"] = (
    medicine_distribution["patient_count"]
    / medicine_distribution["patient_count"].sum()
    * 100
)

medicine_distribution["percentage"] = (
    medicine_distribution["percentage"]
    .round(2)
)

medicine_distribution
```

```python
disease_medicine_relationship_count = (
    disease_medicine_map
    .groupby("disease")
    .agg(
        medicine_class_count=("medicine_class", "nunique")
    )
    .reset_index()
    .sort_values(
        "medicine_class_count",
        ascending=False
    )
)

disease_medicine_relationship_count
```

```python
for disease, group in disease_medicine_map.groupby("disease"):
    
    medicines = group["medicine_class"].tolist()
    
    print(f"\nDisease: {disease}")
    
    for medicine in medicines:
        print(f"  → {medicine}")
```

```python
patient_validation = (
    patient_disease_medicine_unique
    .groupby("patient_id")
    .agg(
        disease_count=("disease", "nunique"),
        medicine_class_count=("medicine_class", "nunique")
    )
    .reset_index()
)

patient_validation.head()
```

```python
patients_without_disease = (
    patient_disease_medicine_unique[
        patient_disease_medicine_unique["disease"].isna()
    ]
)

print(
    "Patients without disease information:",
    len(patients_without_disease)
)
```

```python
patients_without_medicine = (
    patient_disease_medicine_unique[
        patient_disease_medicine_unique["medicine_class"].isna()
    ]
)

print(
    "Patients without medication-class information:",
    len(patients_without_medicine)
)
```

```python
validation_report = pd.DataFrame({
    "metric": [
        "Original Day 7 records",
        "Unique patients",
        "Unique diseases/conditions",
        "Unique medication classes",
        "Unique Patient-Disease relationships",
        "Unique Patient-Disease-Medicine relationships",
        "Duplicate Patient-Disease-Medicine rows",
        "Missing patient IDs",
        "Missing diseases",
        "Missing medication classes"
    ],
    "value": [
        len(patient_disease_medicine),
        patient_disease_medicine_unique["patient_id"].nunique(),
        patient_disease_medicine_unique["disease"].nunique(),
        patient_disease_medicine_unique["medicine_class"].nunique(),
        len(patient_disease),
        len(patient_disease_medicine_unique),
        duplicate_relationships,
        patient_disease_medicine_unique["patient_id"].isna().sum(),
        patient_disease_medicine_unique["disease"].isna().sum(),
        patient_disease_medicine_unique["medicine_class"].isna().sum()
    ]
})

validation_report
```

```python
assert (
    patient_disease_medicine_unique["patient_id"]
    .notna()
    .all()
)

assert (
    patient_disease_medicine_unique["disease"]
    .notna()
    .all()
)

assert (
    patient_disease_medicine_unique["medicine_class"]
    .notna()
    .all()
)

assert (
    patient_disease_medicine_unique
    .duplicated()
    .sum()
    == 0
)

print("All Patient–Disease–Medicine validation checks passed successfully.")
```

```python
print("Final Patient–Disease–Medicine Dataset")
print("=" * 60)

print("Rows:", patient_disease_medicine_unique.shape[0])
print("Columns:", patient_disease_medicine_unique.shape[1])

print("\nColumns:")
print(patient_disease_medicine_unique.columns.tolist())

print("\nData types:")
print(patient_disease_medicine_unique.dtypes)
```

```python
with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    patient_disease_medicine_unique.to_excel(
        writer,
        sheet_name="Patient_Disease_Medicine",
        index=False
    )

    disease_medicine_map.to_excel(
        writer,
        sheet_name="Disease_Medicine_Map",
        index=False
    )

    patient_disease.to_excel(
        writer,
        sheet_name="Patient_Disease",
        index=False
    )

    patient_count_by_disease.to_excel(
        writer,
        sheet_name="Disease_Patient_Summary",
        index=False
    )

    disease_summary.to_excel(
        writer,
        sheet_name="Disease_Summary",
        index=False
    )

    medicine_class_summary.to_excel(
        writer,
        sheet_name="Medicine_Summary",
        index=False
    )

    medicine_distribution.to_excel(
        writer,
        sheet_name="Medicine_Distribution",
        index=False
    )

    disease_medicine_counts.to_excel(
        writer,
        sheet_name="Disease_Medicine_Counts",
        index=False
    )

    validation_report.to_excel(
        writer,
        sheet_name="Validation_Report",
        index=False
    )

print("Day 7 workbook saved successfully:")
print(output_file)
```

```python
if output_file.exists():

    print("Output file exists:", True)
    print("Output file:", output_file)

    saved_excel = pd.ExcelFile(output_file)

    print("\nSaved sheets:")
    print(saved_excel.sheet_names)

else:
    raise FileNotFoundError(
        "Day 7 output workbook was not created."
    )
```
